# Prime Number Code 

```python
n = int(input("Enter a number : "))
flag = True

for i in range(2 , n):
    if n%i == 0:
        flag = False

if flag and n>1:
    print("The num is prime")
else:
    print("The num is Not Prime")
```

# Lambda Functions 

```python
# WAP to take number sequence 10 nums , filter even numbers , square them and find sum using map filter and reduce

from functools import reduce

nums = [12,324,54,45,344,65,687,88,76,624]

even_nums = list(filter(lambda x : x%2 == 0 , nums))
square_nums = list(map(lambda x : x*x , even_nums))
sum_square_nums = reduce(lambda x , y : x+y , square_nums)

print(f"The even numbers are {even_nums}")
print(f"The square of even numbers are {square_nums}")
print(f"The sum of square of even numbers are {sum_square_nums}")
```
