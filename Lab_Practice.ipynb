```python
user_income = float(input("==============Bhai Kitna Kamate H000 ??????????? ============== "))
user_age = int(input("==============Hanji Kitne saal ka tajurba hai aapko ???=============="))

# Income tax calculation based on income slabs

if user_income >= 0 and user_income < 250000:   
    print("No Income tax will be charged")
elif user_income > 250000:
    tax =0 
    if user_income == 250000:
        tax = (user_income * 0.5) - 50000
    elif user_income > 500000 and user_income < 1000000:
        tax = (user_income - 500000) * 0.2 
    elif user_income >= 1000000 and user_income < 3000000:
        tax = (tax*0.07) + tax 
    elif user_income >= 3000000:
        tax = (user_income - 50000) * 0.3 
else :
    print("Invalid input. Please enter a positive number.")

print(f"Your income tax is: {tax}")

# Income Tax based on age slabs

if user_age > 60 :
    reduce_tax = tax -(tax * 0.03) 
    print(f"Your income tax after age reduction is: {reduce_tax}")
```

```python
# WAP to print the total sales taking sales as input in a loop , total will calculate until the user want to break
total_sales=0
while True :
    user_sales = int(input("Enter the user sales here : "))

    if user_sales != 0 :
        total_sales = total_sales + user_sales
    else :
        break

print(f"The total sales of the user is {total_sales}")
```

```python
# WAP for an e-commerce website want to apply a 10% discount to all products , prices using map and lambda function 
print("Sale!! Sale!! Sale!")

prices=[] # empty list to store the prices of the items

items=int(input("Enter the number of items: ")) 

# Loop to take the price of each item as input and store it in the prices list
for i in range(items):
    price=float(input(f"Enter the price of the item {i+1}:"))
    prices.append(price)

# Using map and lambda function to apply a 10% discount to all prices
discounted_prices=list(map(lambda x:x*0.9,prices))
print("The Original Price is:",prices) # Output the original prices
print("The Discounted Price is:",discounted_prices) # Output the discounted prices
```

 WAP were the store want to maintain an immutable list of product id using tuple and perform the following operations
 1. Display all product id
 2. Count the total product
 3. Find if given product is present in inventry or not 

```python
product_id = ("12345", "67890", "54321", "09876") # Tuple to store product IDs

def display():
    print(f"The total product present in the inventry is : {product_id}")

def count():
    print("The total number of products in the inventory is:", len(product_id))

def find():
    search_id = input("Enter the product ID to search: ")
    if search_id in product_id:
        print(f"Product ID {search_id} is present in the inventory.")
    else:
        print(f"Product ID {search_id} is not found in the inventory.")

display()
count()
find()
```
